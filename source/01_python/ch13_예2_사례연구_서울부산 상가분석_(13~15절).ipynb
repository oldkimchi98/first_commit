```python
import folium
part = ['입시교과학원', '태권도/무술학원']
df_m = df_academy[df_academy['상권업종소분류명'].isin(part)].sample(1000, random_state=3)
df_m.shape
```

```python
df_m.head().index
```

```python
# df_m을 지도시각화
lat_mean = df_m['위도'].mean()
long_mean = df_m['경도'].mean()
m = folium.Map(location=[lat_mean, long_mean], zoom_start=12,)
# tiles="cartodb positron" 특정 타일로 변경
m
```
