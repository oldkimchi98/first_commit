## 2-3 맞춤법 검사기
- 네이버 맞춤법 검사기 이용

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from bs4 import BeautifulSoup
import time
```

```python
driver = webdriver.Chrome()
```

```python
# 1. 맞춤법 검사할 파일 읽기
with open('data/ch14_맞춤법테스트.txt', 'r', encoding='UTF8') as fp:
    text = fp.read()
    
# 2. 300자 기준으로 문장 단위로 나누기
ready_text_list = []

while len(text) >= 300:
    temp = text[:300]
    last_dot_index = temp.rfind('.')

    # 300자 안에 마침표가 없을 경우
    if last_dot_index == -1:
        last_dot_index = 299

    ready_text_list.append(text[:last_dot_index + 1])
    text = text[last_dot_index + 1:]

# 남은 텍스트 추가
ready_text_list.append(text)
print([len(read_text) for read_text in ready_text_list])

# 3. 크롬 실행 및 네이버 접속
driver = webdriver.Chrome()
driver.get('https://www.naver.com')
time.sleep(1)

# 4. 네이버에서 맞춤법 검사기 검색
elem = driver.find_element(By.ID, 'query')
elem.send_keys('맞춤법 검사기')
elem.send_keys(Keys.RETURN)
time.sleep(1)

# 5. 맞춤법 검사기 입력창 찾기
textarea = driver.find_element(By.CLASS_NAME, 'txt_gray')
results = ''

# 6. 텍스트를 하나씩 맞춤법 검사
for idx, read_text in enumerate(ready_text_list):
    print(f'검사중... {idx + 1}/{len(ready_text_list)}')
    # 입력창 비우기
    textarea.clear()
    # 검사할 텍스트 입력
    textarea.send_keys(read_text)
    # 맞춤법 검사 버튼
    btn = driver.find_element(By.CLASS_NAME, 'btn_check')
    btn.click()
    time.sleep(1)

    # 페이지 HTML 가져오기
    soup = BeautifulSoup(driver.page_source, 'html.parser')
    # 검사 결과 가져오기
    result = soup.select_one('p._result_text.stand_txt')
    if result:
        results += result.text
    else:
        print('검사 결과를 찾을 수 없습니다.')
    time.sleep(1)

# 7. 브라우저 종료
driver.close()

# 8. 맞춤법 검사 결과 파일 저장
with open('data/ch14_맞춤법후.txt', 'w', encoding='UTF8') as fp:
    fp.write(results)

print('맞춤법 검사 완료')
```

```python
# 맞춤법 검사 결과(results)를 파일 출력
with open('data/ch14_맞춤법후.txt', 'w') as fp: # 인코딩 지정안하면 cp949
    fp.write(results)
```

# 3절. 연습문제
- https://papago.naver.com/ 을 통해서 "data/ch14_맞춤법후.txt"파일의 내용을 영문으로 번역하여 파일 출력

```python
# 1. 네이버 맞춤법 검사기 절차 이후 저장한 맞춤법후 txt 파일 읽어오기
with open('data/ch14_맞춤법후.txt', 'r') as fp:
    text = fp.read()

ready_text_list = []
# 2. 파파고 최대 글자 제한 수 3000자에 맞춰서 while문으로 텍스트 분할
while len(text) >= 3000:
    temp = text[:3000]
    last_dot_index = temp.rfind('.')
    
    if last_dot_index == -1:
        last_dot_index = 2999
    
    ready_text_list.append(text[:last_dot_index+1])
    text = text[last_dot_index+1:]

ready_text_list.append(text)
print([len(read_text) for read_text in ready_text_list])

# 3. 파파고 영문 번역 시작
driver = webdriver.Chrome()
driver.get('https://papago.naver.com/')
time.sleep(1)

# 4. 팝업창 x버튼 위치 지정 후 if문으로 닫음
btn = driver.find_element(By.CLASS_NAME, 'entry-popup-module-scss-module__UZIxta__close')
if btn:
    btn.click()
    print('축하 창 닫음')
else:
    print('축하 창 안뜸')
input_elem = driver.find_element(By.CLASS_NAME,
                                'text-translator-module-scss-module__CYJRkW__text-editor')

# 5. 파파고 번역후 결과 누적
translation_results = '' # 번역결과를 담을 빈 변수

for idx, read_text in enumerate(ready_text_list):
    print(f'번역중...{idx+1}/{len(ready_text_list)}')
    
    input_elem.clear()
    input_elem.send_keys(Keys.CONTROL, 'a')
    input_elem.send_keys(read_text)
    time.sleep(2)
    
    result  = driver.find_elements(By.CSS_SELECTOR, 'div[class^="text-editor-module-scss-module"]')
    
    if len(result) >= 2:
         translation_results += result[1].text + '\n\n'
    else:
         print(f'{idx+1}번째 번역 결과를 찾지 못했습니다.')
    
print(translation_results)
driver.close()

# 파파고 최종 번역본 파일 저장
with open('data/ch14_자동화영어번역본.txt', 'w', encoding = 'UTF8') as fp:
    fp.write(translation_results)
```
