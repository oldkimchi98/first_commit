```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(model='llama3.2:1b')

food_prompt = PromptTemplate(
    template='What is the most famous food in {country}? Return the name of the food only.',
    input_variables=['country']
)
outputParser = StrOutputParser()
food_chain = food_prompt | llm | outputParser

country = input("나라 이름을 입력하세요 : ")
food_chain.invoke({'country': country})
```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatOllama(model='llama3.2:1b')

recipe_prompt = PromptTemplate(
    template='What is the recipe for {food}?',
    input_variables=['food']
)
outputParser = StrOutputParser()
recipe_chain = recipe_prompt | llm | outputParser

food = input("음식 이름을 입력하세요 : ")
recipe_chain.invoke({'food': food})
```
