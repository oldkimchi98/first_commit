# 결측치 확인

```python
# 결측치 확인
def check_missing(df):
    result = pd.DataFrame({
        "결측치": df.isnull().sum(),
        "결측률(%)": (df.isnull().mean() * 100).round(2)
    })
    return result[result["결측치"] > 0]

display(check_missing(store))
display(check_missing(change))
display(check_missing(sales))
display(check_missing(sales_bg))
```

# 기초통계 확인

## describe() 각 행의 의미
- count : 컬럼마다 다르면 그 컬럼에 결측치가 있다는 의미.
- mean vs 50%(중앙값) : 둘이 크게 다르면 값이 한쪽으로 치우침(극단값 존재)
- min/max : 불가능한 값, 극단적인 이상치 확인
- 25%/75% : 대부분의 데이터가 실제로 어디에 모여있는지
- std : 값이 얼마나 들쭉날짝한지

### 빠른 체크
- mean ≈ 50% : 고른 분포, 평균 써도 OK
- mean ≫ 50% : 큰 값 쪽으로 치우침, 중앙값 쓰거나 로그 변환 고려
- mean > 75% : 극단값 영향이 매우 큼
- 25%~75%가 모두 0 : 대부분이 0인 데이터
- std가 mean보다 훨씬 큼	: 값이 매우 들쭉날쭉
- max가 75%의 수백~수천 배 :	이상치 확인 필요
- min/max가 현실적으로 불가능 :	오류 가능성

```python
# 기초통계 확인
display(store.describe())        # 점포-상권
display(change.describe())    # 상권변화지표-상권
display(sales.describe())        # 추정매출-상권
display(sales_bg.describe())  # 추정매출-상권배후지
```

```python
import pandas as pd

# 숫자 하나를 한국식 단위로 바꾸는 함수
def 한국식(x, 소수=1):
    if pd.isna(x):
        return ''
    부호 = '-' if x < 0 else ''
    x = abs(x)
    for 기준, 단위 in [(1e12, '조'), (1e8, '억'), (1e4, '만')]:
        if x >= 기준:
            return f"{부호}{x/기준:,.{소수}f}{단위}"
    return f"{부호}{x:,.0f}" if x == int(x) else f"{부호}{x:,.2f}"

# describe() 결과에서 숫자만 한국식으로 바꾸기 (행·컬럼 이름은 그대로)
def 보기좋은_기초통계(df, 제외=None):
    num = df.select_dtypes('number')
    if 제외:
        num = num.drop(columns=[c for c in 제외 if c in num.columns])
    결과 = num.describe()
    if hasattr(결과, 'map'):          # pandas 2.1 이상
        return 결과.map(한국식)
    return 결과.applymap(한국식)       # pandas 2.1 미만

# 클래스 사용 및 출력
display(보기좋은_기초통계(store, 제외=['stdr_yyqu_cd', 'trdar_cd']))
display(보기좋은_기초통계(change, 제외=['기준_년분기_코드', '상권_코드']))
display(보기좋은_기초통계(sales, 제외=['기준_년분기_코드', '상권_코드']))
display(보기좋은_기초통계(sales_bg, 제외=['stdr_yyqu_cd', 'trdar_cd']))
```

```python
sales_cols = [
    '당월_매출_금액',
    '당월_매출_건수',
    '주중_매출_금액',
    '주말_매출_금액'
]

display(
    sales[sales_cols]
    .describe()
    .T
    .round(2)
)
```
